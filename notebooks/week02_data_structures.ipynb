| b      | float  | No       |

| d      | tuple  | No       |
| e      | list   | Yes      |

| g      | set    | Yes      |

e.append(4) works because lists are mutable — you can change them in place.

f["b"] = 2 works because dictionaries are mutable — you can add or modify entries.

d[0] = 99 fails because tuples are immutable — their contents cannot be changed once created.

```python
def add_value(values):
    values.append(100)

data = [1, 2, 3]
add_value(data)

print(data)
```

```python
def add_value(values):
    result = values.copy()
    result.append(100)
    return result

data = [1, 2, 3]
new_data = add_value(data)

print(data)
print(new_data)
```

```python
states = [
    "Arizona", "Arizona", "Utah",
    "Nevada", "Arizona", "Utah"
]

unique_states = set(states)
```

```python
unique_states
"Arizona" in unique_states
"California" in unique_states
```

A set is useful because it automatically removes duplicates and makes membership testing fast.
When you want unique categories and quick “is this in the group?” checks, a set is the right tool.
