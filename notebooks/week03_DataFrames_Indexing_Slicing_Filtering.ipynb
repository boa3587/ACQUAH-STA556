Reflection
1. df.loc[101]
This selects the row whose index label is 101.

Because you set "id" as the index, 101 is now a label, not a position.

So .loc means:

✔ “Find the row with index label 101.”
2. df.iloc[0]
This selects the first row by position, regardless of what the index label is.

So .iloc means:

✔ “Give me the row in position 0.”

Question
Why does label‑based slicing behave differently from positional slicing?
✔ Because .loc slicing is inclusive, while .iloc slicing is exclusive.
.iloc[start:stop]  
→ uses positions  
→ behaves like normal Python slicing
→ stop is NOT included

.loc[start_label:stop_label]  
→ uses index labels  
→ includes both the start and stop labels  
→ stop IS included
