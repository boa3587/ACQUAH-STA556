Based on uniqueness checks in each table, the likely key columns are:

Students table: student_id

Enrollments table: enrollment_id

Each enrollment record is uniquely identified by its enrollment ID.

Courses table: course_id

Both the students and enrollments tables contain missing values in important fields such as age, GPA, final scores, project scores, and attendance percentages. These internal missing values reduce the completeness of demographic and performance analyses. More importantly, the enrollments table contains student_id = S9999, which does not appear in the students table, and course_id = C999, which does not appear in the courses table. These are foreign‑key mismatches that will produce incomplete rows during merging and indicate referential integrity problems in the raw data.

Part 3 — Joining Tables

```python
# Merge enrollments ↔ students

# Expected relationship: Each student can appear in many enrollments, but each enrollment belongs to exactly one student. This is a one‑to‑many relationship on student_id.
# Merge 1: students ↔ enrollments (OUTER)
merge1 = students.merge(
    enrollments,
    how="outer",
    on="student_id",
    validate="one_to_many",
)


merge1
```

```python
# Merge with courses (many-to-one)

# Expected relationship: Each course can appear in many enrollments, but each enrollment refers to exactly one course. This is a many‑to‑one relationship on course_id.
final = merge1.merge(
    courses,
    how="left",                
    on="course_id",
    validate="many_to_one",
    indicator=True
)

final
```

In this dataset, duplicated rows occur because a single student can appear in multiple enrollment records. When the tables are combined, the student‑level information (such as student_id, first_name, program, advisor, etc.) must be repeated for each enrollment that belongs to that student. For example, S1001 appears in three different enrollment records (E5001, E5002, E5003). As a result, the student’s attributes are duplicated across three rows in the merged dataset. This duplication is expected and correct: each enrollment is a separate observation, and each observation must carry the student’s information.

Part 4 — Missing Data and Reshaping

```python
# Missing counts
missing_counts = final.isna().sum()

# Missing percentages
missing_percent = final.isna().mean() * 100

missing_counts, missing_percent
```

```python
missing_gpa = final.loc[final["gpa"].isna()]
missing_gpa

```

```python
wide_scores = final[[
    "student_id",
    "enrollment_id",
    "course_id",
    "term",
    "midterm_score",
    "final_score",
    "project_score"
]]

wide_scores.head()
```

```python
# Select the columns we want to reshape
score_cols = ["midterm_score", "final_score", "project_score"]
# Melt the dataset from wide → long
long_scores = final.melt(
    id_vars=["student_id", "enrollment_id", "course_id", "term"],
    value_vars=score_cols,
    var_name="assessment_type",
    value_name="score"
)

long_scores.head()

```

Replacing missing values with zero is inappropriate because zero is a meaningful numeric value, not a neutral placeholder. Substituting zero changes the distribution of the variable, artificially lowers averages, and introduces bias into any statistical analysis. It also incorrectly implies that missing scores or missing GPA values represent actual zero performance, which is not true. Missing values should be handled using appropriate methods such as imputation or exclusion depending on the context.
